## Experiment 6: Lists, Tuples and Dictionary

### Aim

To understand the use of lists, tuples and dictionaries in Python.

### Introduction

Lists, tuples and dictionaries are important data structures in Python. Lists are mutable, tuples are immutable, and dictionaries store data in key-value pairs. These structures help organize and manage data efficiently.

## Program 1

Write a Python program to perform basic operations on a list.

```python
numbers = [10, 20, 30, 40]

print("Original List:", numbers)

numbers.append(50)

print("After Appending:", numbers)

numbers.remove(20)

print("After Removing:", numbers)
```

The program demonstrates basic list operations such as adding and removing elements using `append()` and `remove()`.

## Program 2

Write a Python program to find the largest element in a list.

```python
numbers = [25, 18, 42, 67, 12]

largest = max(numbers)

print("Largest Element =", largest)
```

The `max()` function returns the largest value present in the list.

## Program 3

Write a Python program to demonstrate tuple operations.

```python
fruits = ("Apple", "Banana", "Mango", "Orange")

print("Tuple:", fruits)
print("First Fruit:", fruits[0])
print("Number of Fruits:", len(fruits))
```

A tuple stores multiple values and cannot be modified after creation. Elements can be accessed using their index.

## Program 4

Write a Python program to create and display a dictionary.

```python
student = {
    "Name": "Ashutosh",
    "Course": "M.Tech",
    "Semester": 1
}

print(student)

print("Name:", student["Name"])
print("Course:", student["Course"])
```

A dictionary stores information in the form of key-value pairs. Values are accessed using their corresponding keys.

## Program 5

Write a Python program to update a dictionary.

```python
student = {
    "Name": "Ashutosh",
    "Marks": 85
}

student["Marks"] = 92
student["City"] = "Delhi"

print(student)
```

The program updates an existing value and adds a new key-value pair to the dictionary.

## Conclusion

In this experiment, we learned how to use lists, tuples and dictionaries in Python. We performed basic operations such as adding and removing list elements, accessing tuple values, and creating and updating dictionaries.
