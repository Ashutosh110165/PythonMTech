## Experiment 7: Functions

### Aim

To understand the concept of functions in Python and perform different operations using user-defined functions.

### Introduction

A function is a block of reusable code that performs a specific task. Functions help make programs modular, readable, and easier to maintain. Python provides built-in functions as well as user-defined functions.

## Program 1

Write a Python program to create a function that adds two numbers.

```python
def add(a, b):
    return a + b

x = int(input("Enter first number: "))
y = int(input("Enter second number: "))

print("Sum =", add(x, y))
```

The function `add()` accepts two numbers as arguments and returns their sum. The numbers are 10 and 20.

## Program 2

Write a Python program to find the square of a number using a function.

```python
def square(num):
    return num * num

n = int(input("Enter a number: "))

print("Square =", square(n))
```

The function receives a number as input and returns its square. The number is 5.

## Program 3

Write a Python program to check whether a number is even or odd using a function.

```python
def check(number):
    if number % 2 == 0:
        return "Even"
    else:
        return "Odd"

n = int(input("Enter a number: "))

print(check(n))
```

The function checks whether the number is divisible by 2 and returns the appropriate result. The number is 1379.

## Program 4

Write a Python program to find the factorial of a number using a function.

```python
def factorial(n):
    fact = 1

    for i in range(1, n + 1):
        fact *= i

    return fact

num = int(input("Enter a number: "))

print("Factorial =", factorial(num))
```

The function calculates the factorial by multiplying all numbers from 1 to the given number. The number is 5.

## Program 5

Write a Python program to find the largest of three numbers using a function.

```python
def largest(a, b, c):
    return max(a, b, c)

x = int(input("Enter first number: "))
y = int(input("Enter second number: "))
z = int(input("Enter third number: "))

print("Largest Number =", largest(x, y, z))
```

The function uses the `max()` function to return the largest of the three given numbers. The numbers are 45, 90, 1.
