## Experiment 10: Data Analysis and Visualization

### Aim

To perform basic data analysis and visualization using NumPy, Pandas and Matplotlib.

### Introduction

NumPy is used for numerical computations, Pandas is used for handling tabular data, and Matplotlib is used to create charts and graphs. These libraries are widely used in data science and data analysis.

## Program 1

Write a Python program to create a NumPy array and perform basic operations.

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("Array:", arr)
print("Sum =", np.sum(arr))
print("Mean =", np.mean(arr))
print("Maximum =", np.max(arr))
```

This program creates a NumPy array and calculates its sum, mean and maximum value using built-in NumPy functions.

## Program 2

Write a Python program to create a Pandas DataFrame.

```python
import pandas as pd

data = {
    "Name": ["Aman", "Riya", "Karan"],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(data)

print(df)
```

A DataFrame is a two-dimensional table used to store and organize data in rows and columns.

## Program 3

Write a Python program to display statistical information of a DataFrame.

```python
import pandas as pd

data = {
    "Marks": [85, 90, 78, 95, 88]
}

df = pd.DataFrame(data)

print(df.describe())
```

The `describe()` function displays statistical measures such as count, mean, standard deviation, minimum and maximum values.

## Program 4

Write a Python program to draw a bar graph.

```python
import matplotlib.pyplot as plt

subjects = ["Python", "Java", "C++", "DBMS"]
marks = [85, 90, 80, 88]

plt.bar(subjects, marks)

plt.title("Marks in Subjects")
plt.xlabel("Subjects")
plt.ylabel("Marks")

plt.show()
```

The `bar()` function is used to create a bar graph for comparing values.

## Program 5

Write a Python program to draw a line graph.

```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5]
sales = [120, 150, 180, 170, 200]

plt.plot(days, sales)

plt.title("Daily Sales")
plt.xlabel("Day")
plt.ylabel("Sales")

plt.show()
```

The `plot()` function is used to draw a line graph showing the trend of data over time.
