### Explanation

The `print()` function is used to display text or values on the screen. It is one of the most commonly used functions in Python.

```python
student_name = "Ashutosh"
course = "M.Tech"
semester = 1

print("Student Name:", student_name)
print("Course:", course)
print("Semester:", semester)
```

```python
student_name = "Ashutosh"
age = 22
cgpa = 7.7
hosteller = True

print(type(student_name))
print(type(age))
print(type(cgpa))
print(type(hosteller))
```

```python
first_name = "Ashutosh"
last_name = "Das"

print(first_name)
print(last_name)
print(first_name + " " + last_name)
```

```python
name, course, year = "Ashutosh", "M.Tech", 1

print(name)
print(course)
print(year)
```

```python
python_marks = 88
java_marks = 92

print("Before Swapping")
print("Python:", python_marks)
print("Java:", java_marks)

python_marks, java_marks = java_marks, python_marks

print("\nAfter Swapping")
print("Python:", python_marks)
print("Java:", java_marks)
```

```python
student = input("Enter your name: ")
branch = input("Enter your branch: ")

print("Welcome,", student)
print("Branch:", branch)
```

```python
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

print("Addition =", num1 + num2)
print("Subtraction =", num1 - num2)
print("Multiplication =", num1 * num2)

if num2 != 0:
    print("Division =", num1 / num2)
else:
    print("Division by zero is not allowed.")
```
