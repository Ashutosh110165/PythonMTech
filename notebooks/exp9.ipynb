## Experiment 9: Classes and Objects

### Aim

To understand the concept of classes and objects in Python and perform basic object-oriented programming operations.

### Introduction

Object-Oriented Programming (OOP) is a programming approach that uses classes and objects. A class is a blueprint for creating objects, while an object is an instance of a class. OOP helps in organizing code and promotes reusability.

## Program 1

Write a Python program to create a class and an object.

```python
class Student:
    def display(self):
        print("Welcome to Python Programming Lab")

student = Student()

student.display()
```

The program creates a class named `Student`, creates an object of the class, and calls its method using the object.

## Program 2

Write a Python program to use a constructor.

```python
class Student:
    def __init__(self, name):
        self.name = name

    def display(self):
        print("Student Name:", self.name)

student = Student("Ashutosh")

student.display()
```

The `__init__()` method is called automatically when an object is created. It initializes the object's data.

## Program 3

Write a Python program to create a class with multiple attributes.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def details(self):
        print("Name:", self.name)
        print("Age:", self.age)

student = Student("Ashutosh", 21)

student.details()
```

The object stores multiple attributes such as name and age, which are displayed using a class method.

## Program 4

Write a Python program to demonstrate inheritance.

```python
class Animal:
    def sound(self):
        print("Animals make sounds.")

class Dog(Animal):
    def bark(self):
        print("Dog barks.")

dog = Dog()

dog.sound()
dog.bark()
```

Inheritance allows one class to acquire the properties and methods of another class. Here, the `Dog` class inherits from the `Animal` class.

## Program 5

Write a Python program to store and display student details using a class.

```python
class Student:
    def __init__(self, name, roll_no, course):
        self.name = name
        self.roll_no = roll_no
        self.course = course

    def display(self):
        print("Name:", self.name)
        print("Roll No:", self.roll_no)
        print("Course:", self.course)

student = Student("Ashutosh", 101, "M.Tech")

student.display()
```

The program stores student information inside a class and displays it using an object.
