## Experiment 4: Loops

### Aim

To understand the use of looping statements (`for` and `while`) in Python for executing a block of code repeatedly.

### Introduction

Loops are used when a task needs to be repeated multiple times. Python mainly provides two looping statements: `for` and `while`. These loops help reduce code repetition and make programs more efficient.

## Program 1

Write a Python program to print numbers from 1 to 10 using a `for` loop.

```python
for i in range(1, 11):
    print(i)
```

The `for` loop repeats the code for every value in the given range. Here, it prints numbers from 1 to 10.

## Program 2

Write a Python program to display the multiplication table of a given number.

```python
number = int(input("Enter a number: "))

for i in range(1, 11):
    print(number, "x", i, "=", number * i)
```

The program uses a `for` loop to generate the multiplication table from 1 to 10 for the entered number.The number used is 16.

## Program 3

Write a Python program to find the sum of the first n natural numbers.

```python
n = int(input("Enter a number: "))

total = 0

for i in range(1, n + 1):
    total += i

print("Sum =", total)
```

The loop adds each number from 1 to 12 and stores the result in the variable `total`.

## Program 4

Write a Python program to find the factorial of a number.

```python
number = int(input("Enter a number: "))

factorial = 1

for i in range(1, number + 1):
    factorial *= i

print("Factorial =", factorial)
```

The factorial of a number is the product of all positive integers from 1 to that number. The loop multiplies these numbers one by one. The number is 4.

## Program 5

Write a Python program to print all even numbers from 1 to 50.

```python
for i in range(2, 51, 2):
    print(i)
```

The loop starts from 2 and increases by 2 each time, so only even numbers are printed.

## Program 6

Write a Python program to print numbers from 10 to 1 using a `while` loop.

```python
count = 10

while count >= 1:
    print(count)
    count -= 1
```

The `while` loop continues as long as the condition is true. The value of `count` decreases after each iteration until it reaches 1.

## Conclusion

In this experiment, we learned how to use `for` and `while` loops in Python. Loops help automate repetitive tasks such as printing sequences, calculating sums, finding factorials, and generating tables.
