## Experiment 8: File Handling and Exception Handling

### Aim

To understand file handling and exception handling in Python.

### Introduction

File handling allows programs to create, read, write and append data to files. Exception handling helps prevent program crashes by handling runtime errors using `try` and `except` blocks.

## Program 1

Write a Python program to create a file and write data into it.

```python
file = open("sample.txt", "w")

file.write("Welcome to Python Programming Lab.")

file.close()

print("Data written successfully.")
```

The program creates a file named `sample.txt` and writes data into it using write mode (`w`).

## Program 2

Write a Python program to read data from a file.

```python
file = open("sample.txt", "r")

content = file.read()

print(content)

file.close()
```

The program opens the file in read mode (`r`) and displays its contents.

## Program 3

Write a Python program to append data to an existing file.

```python
file = open("sample.txt", "a")

file.write("\nThis is the second line.")

file.close()

print("Data appended successfully.")
```

Append mode (`a`) adds new data at the end of the existing file without deleting previous content.

## Program 4

Write a Python program to handle division by zero using exception handling.

```python
try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))

    print("Result =", a / b)

except ZeroDivisionError:
    print("Division by zero is not allowed.")
```

The `try` block executes the code, and if division by zero occurs, the `except` block handles the error instead of crashing the program. The numbers are 6 and 0.

## Program 5

Write a Python program to handle invalid user input.

```python
try:
    number = int(input("Enter a number: "))

    print("You entered:", number)

except ValueError:
    print("Please enter a valid integer.")
```

If the user enters something other than an integer, the `ValueError` exception is caught and an appropriate message is displayed. The number is 7.

## Conclusion

In this experiment, we learned how to create, read and append data to files. We also learned how exception handling using `try` and `except` makes Python programs more reliable by handling runtime errors gracefully.
