## Experiment 5: Strings and Sets

### Aim

To understand string operations and set operations in Python.

### Introduction

Strings are used to store text, while sets are used to store unique elements. Python provides many built-in functions and methods to perform different operations on strings and sets efficiently.

## Program 1

Write a Python program to find the length of a string.

```python
text = input("Enter a string: ")

print("Length of the string =", len(text))
```

The `len()` function returns the total number of characters present in the given string.
The string is heythere.

## Program 2

Write a Python program to reverse a string.

```python
text = input("Enter a string: ")

print("Reversed String =", text[::-1])
```

The slicing operator `[::-1]` reverses the string by reading it from the last character to the first. The string is agoodday.

## Program 3

Write a Python program to count the number of vowels in a string.

```python
text = input("Enter a string: ")

count = 0

for ch in text.lower():
    if ch in "aeiou":
        count += 1

print("Number of vowels =", count)
```

The program checks every character of the string and increases the count whenever it finds a vowel. The string is nice to meet you.

## Program 4

Write a Python program to find the union of two sets.

```python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

print("Union =", set1 | set2)
```

The union operator (`|`) combines all unique elements from both sets.

## Program 5

Write a Python program to find the intersection of two sets.

```python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

print("Intersection =", set1 & set2)
```

The intersection operator (`&`) returns only the elements that are common to both sets.
