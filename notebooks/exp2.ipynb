## Experiment 2: Input Statements and Operators

### Aim

To understand the use of input statements and different operators in Python.

### Introduction

Python provides the `input()` function to accept data from the user. It also supports various operators such as arithmetic, relational, logical and assignment operators for performing different operations.

## Program 1

Write a Python program to perform arithmetic operations on two numbers.

```python
x = 9
y = 7

print("Addition:", x + y)
print("Subtraction:", x - y)
print("Multiplication:", x * y)
print("Division:", x / y)
print("Modulus:", x % y)
```

## Program 2

Write a Python program to calculate the area of a circle.

```python
import math

radius = float(input("Enter radius: "))

area = math.pi * radius ** 2

print("Area of Circle =", area)
```

Here the radius is 2.5 units 

## Program 3

Write a Python program to calculate (x + y)².

```python
x = 4
y = 3

result = (x + y) ** 2

print("Result =", result)
```

## Program 4

Write a Python program to find the hypotenuse of a right-angled triangle.

```python
import math

base = 3
height = 4

hypotenuse = math.sqrt(base**2 + height**2)

print("Hypotenuse =", hypotenuse)
```

## Program 5

Write a Python program to calculate simple interest.

```python
principal = 10000
rate = 8
time = 2

simple_interest = (principal * rate * time) / 100

print("Simple Interest =", simple_interest)
```
